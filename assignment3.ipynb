```python
import numpy as np
```

```python
#Q1
arr = np.random.randint(1, 101, size=(3, 3))

min_val = np.min(arr)
max_val = np.max(arr)

normalized_arr = (arr - min_val) / (max_val - min_val)
print("Original Array:")
print(arr)

print("\nNormalized Array:")
print(normalized_arr)
```

```python
#Q2

arr = np.random.randint(1, 101, 10)


x = int(input("Enter a value X: "))

closest_value = arr[np.abs(arr - x).argmin()]

print("Array:", arr)
print("Value of X:", x)
print("Closest value:", closest_value)
```

```python
#Q3

arr = np.random.randint(1, 101, size=(5, 5))


second_largest = np.sort(arr, axis=0)[-2]


print("Array:")
print(arr)

print("\nSecond-largest value in each column:")
print(second_largest)
```

```python
#Q4

matrix = np.random.randint(1, 101, size=(6, 4))

row_sums = np.sum(matrix, axis=1)

even_row_indices = np.where(row_sums % 2 == 0)[0]

print("Matrix:")
print(matrix)

print("\nRow sums:")
print(row_sums)

print("\nIndices of rows with even sums:")
print(even_row_indices)
```

```python
#Q5

arr = np.random.randint(1, 51, 30)

arr[(arr % 3 == 0) | (arr % 5 == 0)] = -1

print("Modified Array:")
print(arr)
```

```python
#Q6

arr = np.random.randint(1, 101, size=(3, 3))

row_means = np.mean(arr, axis=1, keepdims=True)

result = arr - row_means

print("Original Array:")
print(arr)

print("\nRow Means:")
print(row_means)

print("\nArray After Subtracting Row Means:")
print(result)
```
