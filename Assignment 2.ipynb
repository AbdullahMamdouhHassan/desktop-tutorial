```python
#Q1
def L(list1, num):
    list2 = []

    for item in list1:
        if len(item) > num:
            list2.append(item)

    return list2


result = L(["apple", "pie", "banana"], 3)
print(result)
```

```python
#Q5
class Restaurant:
    def __init__(self):
        self.menu_items = {}          # item -> price
        self.book_table = {}          # table -> customer
        self.customer_orders = []     # list of orders

    def add_item_to_menu(self, item, price):
        self.menu_items[item] = price

    def book_tables(self, table_number, customer_name):
        self.book_table[table_number] = customer_name

    def customer_order(self, customer_name, order):
        self.customer_orders.append(
            {"customer": customer_name, "order": order}
        )

    def print_menu(self):
        print("Menu:")
        for item, price in self.menu_items.items():
            print(f"{item}: ${price}")

    def print_table_reservations(self):
        print("\nTable Reservations:")
        for table, customer in self.book_table.items():
            print(f"Table {table}: {customer}")

    def print_customer_orders(self):
        print("\nCustomer Orders:")
        for order in self.customer_orders:
            print(f"{order['customer']}: {order['order']}")



restaurant = Restaurant()


restaurant.add_item_to_menu("Pizza", 10)
restaurant.add_item_to_menu("Burger", 8)
restaurant.add_item_to_menu("Pasta", 12)


restaurant.book_tables(1, "Ahmed")
restaurant.book_tables(2, "Sara")


restaurant.customer_order("Ahmed", ["Pizza", "Burger"])
restaurant.customer_order("Sara", ["Pasta"])


restaurant.print_menu()
restaurant.print_table_reservations()
restaurant.print_customer_orders()
```

```python
#Q6
class Customer:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    
    def set_name(self, name):
        self.name = name

    def set_age(self, age):
        self.age = age

    
    def get_name(self):
        return self.name

    def get_age(self):
        return self.age


class Bike:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    
    def set_name(self, name):
        self.name = name

    def set_price(self, price):
        self.price = price

    
    def get_name(self):
        return self.name

    def get_price(self):
        return self.price


class MainBikeRental:
    def __init__(self, ListOfBikes=None, ListOfRentals=None):
        self.ListOfBikes = ListOfBikes if ListOfBikes else []
        self.ListOfRentals = ListOfRentals if ListOfRentals else []

    
    def set_bikes(self, ListOfBikes):
        self.ListOfBikes = ListOfBikes

    def set_rentals(self, ListOfRentals):
        self.ListOfRentals = ListOfRentals

    
    def get_bikes(self):
        return self.ListOfBikes

    def get_rentals(self):
        return self.ListOfRentals

    def requestBike(self, customer, num_of_bikes):

        if customer.get_age() <= 6:
            print("Customer must be older than 6 years.")
            return

        if num_of_bikes <= len(self.ListOfBikes):

            for _ in range(num_of_bikes):
                bike = self.ListOfBikes.pop()
                self.ListOfRentals.append(bike)

            print(f"{num_of_bikes} bike(s) rented successfully.")

        else:
            print("Wait 10 mins")

    def returnBike(self, rentalTime, num_of_bikes):

        bill = rentalTime * 40

        for _ in range(num_of_bikes):
            if self.ListOfRentals:
                bike = self.ListOfRentals.pop()
                self.ListOfBikes.append(bike)

        print("Total bill cost:", bill)
        return bill

    def addBike(self, bike):
        self.ListOfBikes.append(bike)

    def totalCost(self):
        total = 0

        for bike in self.ListOfBikes:
            total += bike.get_price()

        return total


# ==========================
# Testing
# ==========================

bike1 = Bike("Mountain Bike", 3000)
bike2 = Bike("Road Bike", 2500)
bike3 = Bike("City Bike", 2000)

customer1 = Customer("Ahmed", 20)

shop = MainBikeRental()

shop.addBike(bike1)
shop.addBike(bike2)
shop.addBike(bike3)

print("Total cost of bikes:", shop.totalCost())

shop.requestBike(customer1, 2)

print("Available bikes:", len(shop.get_bikes()))
print("Rented bikes:", len(shop.get_rentals()))

shop.returnBike(3, 2)

print("Available bikes after return:", len(shop.get_bikes()))
print("Rented bikes after return:", len(shop.get_rentals()))
```
